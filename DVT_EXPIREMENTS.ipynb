```python
#EXP 1B
# 1b. Exploratory Data Analysis (EDA)
# Data Science Frameworks: CRISP-DM, SEMMA, KDD

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("healthcare-dataset-stroke-data.csv")


# --------------------------------------------------
# 1. UNDERSTAND THE DATA
# --------------------------------------------------

print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)


# --------------------------------------------------
# 2. CHECK MISSING VALUES
# --------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())


# --------------------------------------------------
# 3. CHECK DUPLICATES
# --------------------------------------------------

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# --------------------------------------------------
# 4. STATISTICAL SUMMARY
# --------------------------------------------------

print("\nStatistical Summary:")
print(df.describe())


# --------------------------------------------------
# 5. UNIQUE VALUES
# --------------------------------------------------

print("\nUnique Values:")
for column in df.columns:
    print(column, ":", df[column].nunique())


# --------------------------------------------------
# 6. STROKE DISTRIBUTION
# --------------------------------------------------

print("\nStroke Distribution:")
print(df["stroke"].value_counts())

plt.figure(figsize=(6, 4))

sns.countplot(x="stroke", data=df)

plt.title("Stroke Distribution")
plt.xlabel("Stroke")
plt.ylabel("Number of Patients")

plt.show()


# --------------------------------------------------
# 7. AGE DISTRIBUTION
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(df["age"], bins=20)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()


# --------------------------------------------------
# 8. GLUCOSE LEVEL DISTRIBUTION
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(df["avg_glucose_level"], bins=20)

plt.xlabel("Average Glucose Level")
plt.ylabel("Frequency")
plt.title("Average Glucose Level Distribution")

plt.show()


# --------------------------------------------------
# 9. BMI DISTRIBUTION
# --------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(df["bmi"], bins=20, kde=True)

plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.title("BMI Distribution")

plt.show()


# --------------------------------------------------
# 10. CORRELATION HEATMAP
# --------------------------------------------------

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10, 6))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()


# --------------------------------------------------
# 11. AGE VS GLUCOSE LEVEL
# --------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x="age",
    y="avg_glucose_level",
    hue="stroke",
    data=df
)

plt.title("Age vs Average Glucose Level")
plt.xlabel("Age")
plt.ylabel("Average Glucose Level")

plt.show()


print("\nEDA Completed Successfully!")
```

```python
#EXP 2
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# 1. Load dataset
df = pd.read_csv("healthcare-dataset-stroke-data.csv")

print("Original Data:")
print(df.head())

# 2. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Fill missing BMI values
df["bmi"] = df["bmi"].fillna(df["bmi"].median())

# 4. Check duplicates
print("\nDuplicate Rows:", df.duplicated().sum())

# 5. Remove duplicates
df = df.drop_duplicates()

# 6. Remove unnecessary ID column
df = df.drop("id", axis=1)

# 7. Check data types
print("\nData Types:")
print(df.dtypes)


# ---------------------------------------------------
# 8. VISUALIZE ORIGINAL DATA
# ---------------------------------------------------

plt.figure(figsize=(10, 5))

plt.hist(df["age"], bins=20, alpha=0.7, label="Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution - Before Scaling")
plt.legend()

plt.show()


# ---------------------------------------------------
# 9. Min-Max Scaling
# ---------------------------------------------------

scaler = MinMaxScaler()

df[["age", "avg_glucose_level", "bmi"]] = scaler.fit_transform(
    df[["age", "avg_glucose_level", "bmi"]]
)

print("\nAfter Min-Max Scaling:")
print(df[["age", "avg_glucose_level", "bmi"]].head())


# ---------------------------------------------------
# 10. VISUALIZE AFTER MIN-MAX SCALING
# ---------------------------------------------------

plt.figure(figsize=(10, 5))

plt.hist(df["age"], bins=20, alpha=0.7)

plt.xlabel("Scaled Age")
plt.ylabel("Frequency")
plt.title("Age Distribution - After Min-Max Scaling")

plt.show()


# ---------------------------------------------------
# 11. Standardization
# ---------------------------------------------------

standard_scaler = StandardScaler()

df[["age", "avg_glucose_level", "bmi"]] = standard_scaler.fit_transform(
    df[["age", "avg_glucose_level", "bmi"]]
)

print("\nAfter Standardization:")
print(df[["age", "avg_glucose_level", "bmi"]].head())


# ---------------------------------------------------
# 12. VISUALIZE AFTER STANDARDIZATION
# ---------------------------------------------------

plt.figure(figsize=(10, 5))

plt.hist(df["age"], bins=20, alpha=0.7)

plt.xlabel("Standardized Age")
plt.ylabel("Frequency")
plt.title("Age Distribution - After Standardization")

plt.show()


# ---------------------------------------------------
# 13. BOXPLOT
# ---------------------------------------------------

plt.figure(figsize=(10, 5))

df[["age", "avg_glucose_level", "bmi"]].boxplot()

plt.title("Boxplot After Standardization")
plt.ylabel("Standardized Values")

plt.show()


# ---------------------------------------------------
# 14. FINAL DATASET INFORMATION
# ---------------------------------------------------

print("\nFinal Dataset:")
df.info()

print("\nData Cleaning Completed!")
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# 1. Load dataset
df = pd.read_csv("healthcare-dataset-stroke-data.csv")

# 2. Check missing values
print("Missing Values:")
print(df.isnull().sum())

# 3. Fill missing BMI values with median
df["bmi"] = df["bmi"].fillna(df["bmi"].median())

# 4. Check and remove duplicates
print("\nDuplicate Rows:", df.duplicated().sum())
df = df.drop_duplicates()

# 5. Remove unnecessary ID column
df = df.drop("id", axis=1)

# 6. Display data types
print("\nData Types:")
print(df.dtypes)

# 7. Min-Max Scaling
minmax = MinMaxScaler()

df[["age", "avg_glucose_level", "bmi"]] = minmax.fit_transform(
    df[["age", "avg_glucose_level", "bmi"]]
)

print("\nAfter Min-Max Scaling:")
print(df[["age", "avg_glucose_level", "bmi"]].head())

# 8. Standardization
standard = StandardScaler()

df[["age", "avg_glucose_level", "bmi"]] = standard.fit_transform(
    df[["age", "avg_glucose_level", "bmi"]]
)

print("\nAfter Standardization:")
print(df[["age", "avg_glucose_level", "bmi"]].head())

# 9. Final dataset information
print("\nFinal Dataset:")
df.info()

print("\nData Cleaning and Transformation Completed!")
```
